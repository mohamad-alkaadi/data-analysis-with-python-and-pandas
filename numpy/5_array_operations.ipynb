```python
import numpy as np
rng = np.random.default_rng(616)
```

```python
sales = [[0, 5, 155, 0, 518], [0, 187, 616,317, 325]]
sales_array = np.array(sales)
sales_array
```

```python
# we can perform an operation on all elements of the array  in numpy
sales_array + 2
```

```python
quantity = sales_array[0,:]
quantity
```

```python
price = sales_array[1,:]
price
```

```python
# multiply all the items in the arrey quantitiy
# with all in price in each index
quantity * price
```

```python
price = (rng.random(10) * 10).round(2)
price
```

```python
inverntory = rng.integers(0, 100, 10)
inverntory
```

```python
price * inverntory
```

```python
# sum all elements in the array
(price * inverntory).sum()
```
