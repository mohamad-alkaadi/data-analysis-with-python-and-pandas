```python
import numpy as np
import pandas as pd

retail_df = pd.read_csv(
    "./retail_2016_2017.csv", skiprows=range(1, 11000), nrows=1000
)

family_array = np.array(retail_df["family"])
sales_array = np.array(retail_df["sales"])
```

```python
filterd_array = sales_array[family_array == "PRODUCE"]
filterd_array
```

```python
rng = np.random.default_rng(2022)
random_array = rng.random(30)
random_array
```

```python
sampled_array = filterd_array[random_array < 0.5]
sampled_array
```

```python
sampled_array_mean = np.mean(sampled_array)
sampled_array_mean
```

```python
sampled_array_median = np.median(sampled_array)
sampled_array_median
```

```python
np.where(
    sampled_array < sampled_array_median,
    "Below_both",
    np.where(sampled_array > sampled_array_mean, "above_both", "above_median"),
)
```
