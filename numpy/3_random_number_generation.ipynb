```python
from numpy.random import default_rng
```

```python
# Computers can't actually generate true random numbers. Instead,
# they use a mathematical formula to create long lists of numbers that look random
# Imagine NumPy keeps a massive, secret book containing millions of random numbers on numbered pages.
# The seed is just the page number you tell NumPy to open to.
```

```python
# here we will use the seed 32 to open to page 32
rng = default_rng(32)
```

```python
# we ask to get a random number between 0 and 1
random_array = rng.random(10)
random_array
```

```python
# return a n number from a normal distribution with a given mean and standerd deviation
# normal(mean, standerd deviation, how many number do we want)
# random numbers that are centered around 5 and standerd deviation of 1 (valus withn the first standerd deviation)
mean, stddev = 5, 1
random_normal = rng.normal(mean, stddev, size=10)
random_normal
```

```python
# genrate random integers to customers between 0(inc) and 10(inc)
# integers.(starting_int, ending_int, num_of_items)
rng.integers(0, 10, 100)
```
