```python
import numpy as np
rng = np.random.default_rng(2022)
```

```python
prices = np.array([5.99 , 6.99, 22.49, 99.99, 4.99, 49.99])
total = prices + 5
total
```

```python
discount_pct = rng.random(6)
discount_pct
```

```python
pct_owed = 1 - discount_pct
pct_owed
```

```python
final_owed = (total * discount_pct).round(2)
final_owed
```
