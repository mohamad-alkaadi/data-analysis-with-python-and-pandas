```python
import numpy as np
```

```python
prices = np.array([5.99, 6.99, 22.49, 99.99, 4.99, 49.99])
prices
```

```python
prices.sort()
prices
```

```python
biggest_3_prices = prices[-3:]
biggest_3_prices
```

```python
print("Mean: ", np.mean(biggest_3_prices))
print("Min: ", np.min(biggest_3_prices))
print("Max: ", np.max(biggest_3_prices))
print("Median: ", np.median(biggest_3_prices))
```
