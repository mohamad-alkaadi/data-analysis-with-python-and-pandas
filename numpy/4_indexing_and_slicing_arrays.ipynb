```python
import numpy as np
```

```python
# to access elements of an array we need to either index them or slice them
# slicing to access a series of elements (Stop is not inclusive)
```

```python
product_array = np.array([
    'fruits', 'vegetables', 'cereal', 'dairy', 'eggs',
    'snacks', 'beverages', 'coffee', 'tea', 'spices'
])
```

```python
# indexing
print(product_array[1])
print(product_array[-1])
```

```python
# slicing
# [start(inc): end(not inc): step_size]
product_array[0:5]
```

```python
# from start to index 5 ([5] not included)
product_array[:6]
```

```python
# from the index of 5 to the end (inc) step size 2
product_array[5::2]
```

```python
# ---------------------- indexing and slicing 2 dim arrays ----------------------
```

```python
product_array_2d = np.array([
    'fruits', 'vegetables', 'cereal', 'dairy', 'eggs',
    'snacks', 'beverages', 'coffee', 'tea', 'spices'
]).reshape(2, 5)
product_array_2d
```

```python
# indexing
product_array_2d[1,2]
```

```python
# slicing array[start_row:end_row:step_row, start_col:end_col:step_col]
product_array_2d[0:2, 0:3]
```

```python
# all rows from col[2] to end (inc)
product_array_2d[:, 2:]
```

```python
# from row[1] all col

product_array_2d[1:, :]
```
