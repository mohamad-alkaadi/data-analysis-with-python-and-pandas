```python
import numpy as np
```

```python
product_array = np.array(['fruits', 'vegetables', 'cereal', 'dairy', 'eggs'])
product_array
```

```python
inventory_array = np.array([12, 102, 18, 0, 0])
inventory_array
```

```python
# where(condition, true case, false case)
np.where(inventory_array <= 0, "Out of Stock", "In Stock")
```

```python
# if the conditon is correct put "out of stock" and if false return product_array elemnts
np.where(inventory_array <= 0, "Out of stock", product_array)
```

```python
my_array = np.arange(20)
my_array
```

```python
np.where(my_array % 2, "even", "odd")
```

```python
np.where(my_array % 2, "even", my_array)
```

```python
# we can do multiple where inside each other
# example: return everything in even and odd except of the number 9
np.where(my_array % 2 == 0, "even", np.where(my_array == 9, my_array, 'odd'))
```
