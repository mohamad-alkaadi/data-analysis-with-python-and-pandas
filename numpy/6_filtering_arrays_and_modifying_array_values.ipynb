```python
import numpy as np
```

```python
sales_array = np.array([[0, 5, 155, 0, 518], [0, 1827, 616, 317, 325]])
sales_array
```

```python
#  we perform a logical test on np array to return a boolean array
sales_array != 0
```

```python
# indexing an array with aboolean array returns an array with elements
# where the boolean value is True
sales_array[sales_array != 0]
```

```python
# multiple logical tests can be combined using:
# | for or
# & for and
sales_array[(sales_array == 616) | (sales_array < 100)]
```

```python
sales_array[(sales_array > 100) & (sales_array < 500)]
```

```python
# store complex filtering criteria in a varible (known as a Boolean mask)
mask = (sales_array > 100) & (sales_array < 500)
sales_array[mask]
```

```python
# we can filter arrays based on values in other arrays
product_array = np.array(['fruits', 'vegetables', 'cereal', 'dairy', 'eggs'])
product_array
```

```python
# we can assign a single value vie indexing
sales_array_fst_row = sales_array[0,:]
sales_array_fst_row
```

```python
sales_array_fst_row[1] = 25
sales_array_fst_row
```

```python
# we can filter and array and change values based on filter
sales_array_fst_row[sales_array_fst_row == 0] = 5
sales_array_fst_row
```

```python
my_array = np.arange(20)
my_array
```

```python
# create an array with all the values inside it is even and odd and put each one 10 times
even_odd = np.array(['even', 'odd'] * 10)
even_odd
```

```python
even_odd != 'odd'
```

```python
my_array[even_odd != 'odd']
```

```python
# change all even numbers to 0
my_array[even_odd != 'odd'] = 0
my_array
```

```python
# return all even and odd numbers
even_odd_mask = (even_odd != 'odd') | (even_odd != 'even')
my_array[even_odd_mask]
```
