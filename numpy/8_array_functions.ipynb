```python
sales_array = np.array([[0, 5, 155, 0, 518],[0, 1827, 616, 317, 325]])
```

```python
# the sort method it sorts the elemets by axis (axis = 1 [row by row] by defult)
# the sort method in numpy perform in-place sorting
# When you use the .sort() method on a NumPy array object,
# it modifies the original array directly and returns None.
# You do not need to assign it to a new variable.
# if you dont wat to sort in place we use new_array = sort(name_of_array)
sales_array.sort()
sales_array
```

```python
# percentile() is a NumPy function used to compute the p-th percentile of the data along a specified axis.
# first parameter is and array and the secound is A percentile tells you what value sits at a specific percentage point in that line.
# percentile sorts the items small to large
# to find the positioin we use this fornula: Position = p/100 * (N - 1)
# example if we want to find 90% ===> Position = 90/100 * (5 - 1) = 3.6
# we have 5 from 0 to 4 indexes and 3.6 is between 3 and 4
# the 3rd index is 40 and the 4th index is 50
# we divide 3.6 to two parts 3 and 0.6
# 3 is the 3rd position which is 40
# 0.6 is the distance between 40 and 50 so
# we subtract them 50 - 40 = 10 and multiply the result with the other part of the position 0.6 * 10 = 6
# then we add the posiotion number we got from 3 to the number we got from 0.6 and that gets us 40 + 0.6 = 40.6
np.percentile([10, 20, 30, 40, 50], 90)
```
