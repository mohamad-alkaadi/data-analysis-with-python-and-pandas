```python
import numpy as np
import pandas as pd
```

```python
# the index in pandas alows  us to access the rows in Pandas series or dataframe
pd_series = pd.Series([0, 5, 155, 0, 518])
pd_series.name = "Sales"
pd_series
```

```python
# we can index the Sries like numpy
pd_series[2]
```

```python
# we can slice the Sries like numpy
pd_series[2:4]
```

```python
# pandas allows the creation of custom index for accessing rows
sales = pd.Series([0, 5, 155, 0, 518])
sales
```

```python
# note it is not recoumended to use categorical indexes
items_index = ["coffee", "bananas", "tea", "coconut", "suger"]
sales.index = items_index
sales
```

```python
sales["tea"]
```

```python
# when we slice with categorical indexes the end value is included
sales["bananas":"coconut"]
```

```python
my_series = pd.Series(range(5), index=["Day 0", "Day 1", "Day 2", "Day 3", "Day 4"])
my_series
```

```python
my_series["Day 2":"Day 4"]
```
