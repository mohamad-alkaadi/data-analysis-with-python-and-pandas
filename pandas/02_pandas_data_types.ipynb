```python
import numpy as np
import pandas as pd
```

```python
sales = [   0,    5,  155,    0,  518,    0, 1827,  616,  317,  325]
sales_series = pd.Series(sales)
sales_series
```

```python
# you can convert to a diffrent data type using astype
sales_series.astype(float)
```

```python
# if we convert to bool if the value is 0 ==> false and if anything else ==> true
sales_series.astype(bool)
```

```python
# if we try to convert to datetime it is not compatible
sales_series.astype("datetime64")
```

```python
sales_series.astype("object")
```

```python
sales_series.astype("string")
```
