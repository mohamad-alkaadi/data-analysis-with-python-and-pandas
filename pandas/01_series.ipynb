```python
import numpy as np
import pandas as pd
```

```python
# the deffrance between pandas series and numpy arrays is that
# series contain an index and an optional name in addition of the array of data
```

```python
# naming the sereies is !-- optional --!
sales = [0, 5, 155, 0, 518,0, 1827,616, 317, 325]
sales_series = pd.Series(sales, name="Sales")
sales_series
```

```python
# key properties of series:
# values ---> the data array in the series
# index  ---> the index array in the series
# name   ---> the optional name for the series
# dtype  ---> the data type of the elemts in values array
# pandas combine a values and index array within one object
```

```python
sales_series.values
```

```python
sales_series.index
```

```python
sales_series.name
```

```python
sales_series.dtype
```

```python
array = np.arange(5)

series = pd.Series(array)
```
