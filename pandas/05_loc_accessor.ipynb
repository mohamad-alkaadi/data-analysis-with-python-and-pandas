```python
import numpy as np
import pandas as pd
```

```python
# .loc is the preferred way to access values by their custom labels
# it works with dataframes and series
# it works with custom index
# .loc[row label, column label]
#  row label:
#     - [single row]    ex: ["pizza"]
#     - [multiple rows] ex: ["mike", "ike"]
#     - [range of rows] ex: ["jan": "dec"] note: includes the end
#
#  column label:
#     the custom column index for the value(s) you want to access

# .loc also work with integar indexes(means it works with both) so it is pref
# if you work with custom integers the slice will work based on the custom integers
#  not based thir posion
# ex: [0:213, 15: 2341, 4: 451, 24: 2412] if you make .loc[0:15] the return is 213, 2341
```

```python
pd_series = pd.Series([0, 5, 155, 0, 518])
items_index = ["coffee", "bananas", "tea", "coconut", "suger"]
pd_series.index = items_index
pd_series
```

```python
pd_series.loc["tea"]
```

```python
pd_series.loc["bananas": "coconut"]
```

```python
int_pd_series = pd.Series(range(5, 10))
int_pd_series
```

```python
int_pd_series.loc[2]
```

```python
# example on custom ints
my_custom_int_series = pd.Series([213,2341,451,2412], index=[0,15,4,24])
my_custom_int_series.loc[0:15]
```

```python
# it wont accept normal index
my_custom_int_series.loc[0:2]
```

```python
# in comparison the normal slice (it sees that we are reaching the 15 index)
my_custom_int_series[0:15]
```

```python
my_series = pd.Series(range(5), index=["Day 0", "Day 1", "Day 2", "Day 3", "Day 4"])
my_series
```

```python
my_series.loc["Day 2"]
```

```python
# to reset an index to integar
reseted_index = my_series.reset_index(drop=True)
reseted_index
```
