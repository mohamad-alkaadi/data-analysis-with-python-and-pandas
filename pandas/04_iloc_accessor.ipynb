```python
import numpy as np
import pandas as pd
```

```python
# the .iloc[] method is the preferred way to access values by their positional index
# it supports non-integer indexes
# more efficient than slicing
# applies to Series and DataFrames
# .iloc[row position, column position]
# row position:
#  - [single row] ex: [0]
#  - [multiple rows] ex: [5, 9]
#  - [range of rows] ex: [0: 11] note: end is not encluded
# column position:
#  - the column position of the value you want to access
```

```python
pd_series = pd.Series([0, 5, 155, 0, 518])
items_index = ["coffee", "bananas", "tea", "coconut", "suger"]
pd_series.index = items_index
pd_series
```

```python
# you can use the position of the series even if you used a custom index
pd_series.iloc[2]
```

```python
# the range will exclude the the end
pd_series.iloc[2:4]
```

```python
# it doesn't support custom indexing
pd_series.iloc["tea"]
```

```python
my_series = pd.Series(range(5), index=["Day 0", "Day 1", "Day 2", "Day 3", "Day 4"])
my_series
```

```python
# get the first row
my_series.iloc[0]
```

```python
# get the last row
my_series.iloc[-1]
```

```python
# you can pass in multiple indexes
my_series.iloc[[1, 3, 4]]
```
