```python
import numpy as np
import pandas as pd
```

```python
# it is possible to have duplicate index values in pandas series or dataframe
# but it is not idvisable to do so
# we can reset the index to solve this issue
# and becouse pandas does not know what to do with our index it will create a
# new column for the new index and it create a column for our previous index which
# andit stores the index to avoid the lost of data (and it becoums a dataframe)
sales = [0, 5, 155, 0, 518]
items_index = ['coffee', 'coffee', 'tea', 'coconut', 'suger']
sales_series = pd.Series(sales, index=items_index, name="Sales")
sales_series
```

```python
sales_series["coffee"]
```

```python
# example on the rest without a drop
reset_index_without_drop = sales_series.reset_index()
reset_index_without_drop
```

```python
type(reset_index_without_drop)
```

```python
# if we set drop=True we remove the previous index (it stais as series)
reset_index_with_drop = sales_series.reset_index(drop=True)
reset_index_with_drop
```

```python
type(reset_index_with_drop)
```
